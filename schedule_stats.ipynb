```python
import pandas as pd                    

schedule_data = pd.read_csv(r'C:\Users\User\Desktop\Class Stuff\Python Practice\Schedule\tests\data\schedule_example.csv')
schedule_data = schedule_data.drop(columns=['Unnamed: 0'])
schedule_data
```

```python
# First, split the data into primary and secondary shifts
primary_shifts = schedule_data[schedule_data['Role'] == 'Primary']
secondary_shifts = schedule_data[schedule_data['Role'] == 'Secondary']

# Then rename the columns, keep only necessary data, and reset the index for the merge of primary shift data
rename_primary = primary_shifts.rename(columns={'RA':'Primary RA', 'Preference': 'Primary Preference'})
primary_shifts_merge = rename_primary[['Week', 'Day', 'Primary RA', 'Primary Preference']].reset_index()

# Repeat for secondary shifts # NOTE : only need Name and preference, We're using indexes to merge
rename_secondary = secondary_shifts.rename(columns={'RA':'Secondary RA', 'Preference': 'Secondary Preference'})
secondary_shifts_merge = rename_secondary[['Secondary RA', 'Secondary Preference']].reset_index()

# Merge the primary and secondary shift dataframes
schedule = pd.merge(left=primary_shifts_merge, right=secondary_shifts_merge, left_index=True, right_index=True)
schedule.drop(columns=['index_x', 'index_y'], inplace=True)
schedule
```
